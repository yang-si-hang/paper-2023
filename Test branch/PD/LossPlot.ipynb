This file plots the loss function with respect to the nodes positions.
L = 4*(A^2) / c^3

```python
import numpy as np
import matplotlib.pyplot as plt


A, B, C = np.array([[0., 0.],
                    [0.5, 0.5],
                    [1., 0.]])

deltas = np.linspace(1, 0, 100, endpoint=False)
```

```python
# 顶角不变
loss_list = []
for delta in deltas:
    # A_tmp = A
    A_tmp = A + delta * np.array([-0.5, -0.5])
    B_tmp = B
    C_tmp = C + delta * np.array([0.5, -0.5])

    # 计算三边的长度
    a = np.linalg.norm(B_tmp - C_tmp)
    b = np.linalg.norm(C_tmp - A_tmp)
    c = np.linalg.norm(A_tmp - B_tmp)

    # 计算三角形的半周长
    s = (a + b + c) / 2

    # 计算三角形的面积平方
    area2 = s * (s - a) * (s - b) * (s - c)

    loss_tmp = 4*area2/np.power(b, 3)

    loss_list.append(loss_tmp)

plt.plot(deltas, loss_list)
plt.show()
```

```python
# 增加三角形高的长度
loss_list = []
for delta in deltas:
    A_tmp = A
    B_tmp = B + delta * np.array([0., 0.5])
    C_tmp = C

    # 计算三边的长度
    a = np.linalg.norm(B_tmp - C_tmp)
    b = np.linalg.norm(C_tmp - A_tmp)
    c = np.linalg.norm(A_tmp - B_tmp)

    # 计算三角形的半周长
    s = (a + b + c) / 2

    # 计算三角形的面积平方
    area2 = s * (s - a) * (s - b) * (s - c)

    loss_tmp = 4*area2/np.power(b, 3)

    loss_list.append(loss_tmp)

plt.plot(deltas, loss_list)
plt.show()
```

```python
# 增加底边的长度
loss_list = []
for delta in deltas:
    A_tmp = A
    B_tmp = B
    C_tmp = C + delta * np.array([1., 0.])

    # 计算三边的长度
    a = np.linalg.norm(B_tmp - C_tmp)
    b = np.linalg.norm(C_tmp - A_tmp)
    c = np.linalg.norm(A_tmp - B_tmp)

    # 计算三角形的半周长
    s = (a + b + c) / 2

    # 计算三角形的面积平方
    area2 = s * (s - a) * (s - b) * (s - c)

    loss_tmp = 4*area2/np.power(b, 3)

    loss_list.append(loss_tmp)

plt.plot(deltas, loss_list)
plt.show()
```
