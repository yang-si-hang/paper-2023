```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
itr_num = 100
gamma = 1.
learning_rate = 0.5
J_true = np.array([[0.8, 0.1,],
                  [0.07, 0.6]])
deform_error_init = np.array([1., 0.5])
a_init = np.ones((4, 1))

W = np.zeros((2, 4))
a = a_init
deform_error = deform_error_init
for itr in range(itr_num):
    action = learning_rate * np.linalg.inv(J) @ deform_error
    deform = J_true @ action
    deform_error = deform_error - deform

    for i in range(2):
        W[i, i*2:(i+1)*2] = action

    error = W @ a - deform
    a_tmp = a - gamma * W.T @ error
    a = a_tmp
    J = a.reshape(2, 2)
```
