```python
pos1 = np.loadtxt("pos32_1.csv", delimiter=",")
pos2 = np.loadtxt("pos32_2.csv", delimiter=",")
pos3 = np.loadtxt("pos32_3.csv", delimiter=",")
pos4 = np.loadtxt("pos32_4.csv", delimiter=",")
```

```python
fig, axs = plt.subplots(3, 1, figsize=(10, 12))

# x分量
axs[0].plot(pos1[:, 0], label='pos1')
axs[0].plot(pos2[:, 0], label='pos2')
axs[0].plot(pos3[:, 0], label='pos3')
axs[0].plot(pos4[:, 0], label='pos4')
axs[0].set_title('X')
axs[0].legend()

# y分量
axs[1].plot(pos1[:, 1], label='pos1')
axs[1].plot(pos2[:, 1], label='pos2')
axs[1].plot(pos3[:, 1], label='pos3')
axs[1].plot(pos4[:, 1], label='pos4')
axs[1].set_title('Y')
axs[1].legend()

# z分量
axs[2].plot(pos1[:, 2], label='pos1')
axs[2].plot(pos2[:, 2], label='pos2')
axs[2].plot(pos3[:, 2], label='pos3')
axs[2].plot(pos4[:, 2], label='pos4')
axs[2].set_title('Z')
axs[2].legend()

plt.tight_layout()
plt.show()
```
