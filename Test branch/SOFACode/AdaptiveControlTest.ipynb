```python
import numpy as np

factor = 1.e0
learning_rate = 1.e0
j_model = np.array([[1, 0],
                    [0, 0.8]])

deseired_x = np.array([1, 2])
x = np.array([0., 0.])
u = np.array([0., 0.])
a = np.array([0.2, 0, -0.1, 0.1])


for itr in range(100):
    j_model = j_model + np.random.rand(2,2)*0.05
    delta_x = j_model @ u
    x = x + delta_x
    e_pos = deseired_x - x

    W = np.array([[u[0], u[1], 0, 0],
                  [0, 0, u[0], u[1]]])
    e = W @ a - delta_x
    a_update = a - factor*W.transpose()@e
    a = a_update
    j_est = np.reshape(a, (2, 2))

    dx = -2*(deseired_x-x)
    du = dx @ j_est
    delta_u_length = np.linalg.norm(learning_rate*du)
    if delta_u_length < 0.1:
        u = -learning_rate*du
    else:
        u = -learning_rate*du*(0.1/delta_u_length)
    print('position error:', e_pos)
    print('j error:', e)
    print('action:', u)
    print('estimated j:', a)
    print('u length:', delta_u_length)
```
