```python
import numpy as np
import matplotlib.pyplot as plt

A, B, C = np.array([[0, 1],
                    [0.5, 0.5],
                    [1, 0]])
delta = np.linspace(1, 0, 100, endpoint=False)

cur_list = []

for delta_tmp in delta:
    B_tmp = B + delta_tmp * np.array([0.1, 0.1])

    # 计算三边的长度
    a = np.linalg.norm(B_tmp - C)
    b = np.linalg.norm(C - A)
    c = np.linalg.norm(A - B_tmp)

    # 计算三角形的半周长
    s = (a + b + c) / 2

    # 计算三角形的面积
    area = np.sqrt(s * (s - a) * (s - b) * (s - c))

    # 计算外接圆的半径
    radius = a * b * c / (4 * area)

    cur_tmp = np.linalg.norm(b)/(2*radius)
    cur_list.append(cur_tmp)
```

```python
# Plotting
plt.plot(delta, cur_list)
plt.show()
```

- 根据这个loss图像，可以看到，中间的点对于曲率变小的影响是合适的，梯度并没有出现loss变小减小的情况。
- 根据可以猜测到曲率为0时候的梯度是一个固定值
