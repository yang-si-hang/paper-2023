```python
"""
使用Broyden方法来估计变形雅可比矩阵
"""
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
itr_num = 100
J_true = np.array([[0.8, 0.1,],
                  [0.07, 0.6]])
deform_error_init = np.array([1., 0.5])

J0 = np.eye(2)*0.1

J = J0
deform_error = deform_error_init
for i in range(itr_num):
    action = 0.1 * np.linalg.inv(J) @ deform_error_init
    delta_deform = J_true @ action
    deform_error = deform_error - delta_deform

    J_tmp = J + np.outer((delta_deform - J @ action), action) / np.dot(action, action)
    J = J_tmp
    J_error = np.linalg.norm(J - J_true)
    print(f"Iteration {i+1}: Jacobian error = {J_error}. Jacobian = {J.flatten()}. Deform error = {deform_error.flatten()}")
```
