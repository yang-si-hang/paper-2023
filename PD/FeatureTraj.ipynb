```python
import numpy as np
import matplotlib.pyplot as plt

feature_pos = np.loadtxt('feature_pos.csv', delimiter=',')

N = feature_pos.shape[0]

for i in range(N):
    pos1, pos2, pos3 = feature_pos[i, :].reshape(3, 2)
    origin = pos2.copy()
    pos1_show = pos1 - origin
    pos2_show = pos2 - origin
    pos3_show = pos3 - origin
    feature_show = np.array([pos1_show, pos2_show, pos3_show])
    
    plt.scatter(feature_show[:,0], feature_show[:,1], color='red', s=100, marker='.')
    
    plt.savefig(f'Figure/feature_pos{i}.png', bbox_inches='tight')
    
plt.close()
```
